### Architecture simplifiée de Spark
L'architecture de Spark repose sur un modèle Maître/Esclaves (Master/Workers) :
* **Master Node** : Contient le *Driver Program* et le *Spark Context*.
* **Cluster Manager** : Gère l'allocation des ressources (Spark Standalone, YARN, Kubernetes).
* **Worker Nodes** : Hébergent les *Executors*, les tâches (*Tasks*) et le cache.

| Composant | Fonction |
| :--- | :--- |
| **Driver** | Le Driver pilote l’application Spark. Il : crée les traitements ; attribue les tâches ; récupère les résultats. |
| **Executors** | Ils exécutent les traitements. Ils travaillent en parallèle sur différentes partitions des données. |
| **Cluster Manager** | Le Cluster Manager attribue les ressources. Exemples : Standalone ; YARN ; Kubernetes. |

# PARTIE 2 — Spark SQL & Analytics Distribuée

## 1. Création de vues temporaires
L'utilisation de Spark SQL permet d'utiliser un langage SQL classique, de simplifier les analyses et de faciliter les jointures. Pour exécuter des requêtes SQL, nous devons d'abord enregistrer nos DataFrames sous forme de vues temporaires.

````python
assets_df.createOrReplaceTempView("assets")
````

## 2. Première requête SQL
Voici un exemple de requête effectuant un regroupement et un comptage :

````python
spark.sql("""
    SELECT asset_type, COUNT(*) as total
    FROM assets
    GROUP BY asset_type
""").show()
````

### Notions de cours importantes (Mécanisme du GROUP BY)
Lorsque vous effectuez un `GROUP BY`, Spark applique les étapes suivantes de manière distribuée :
$$\text{GROUP BY} \longrightarrow \text{Redistribution des données} \longrightarrow \text{SHUFFLE (échange réseau)} \longrightarrow \text{Agrégation finale}$$

## 3. Jointures distribuées
Exemple de requête effectuant des jointures complexes pour analyser l'activité par ville :

````python
spark.sql("""
SELECT 
    e.event_type,
    n.city,
    COUNT(*) as total
FROM mobility_events e
JOIN assets a ON e.asset_id = a.asset_id
JOIN service_nodes n ON a.node_id = n.node_id
GROUP BY e.event_type, n.city
""").show()
````

**Explication pédagogique :** Cette requête récupère les événements, retrouve l’équipement associé, puis la station associée afin de calculer la répartition des événements par ville.
